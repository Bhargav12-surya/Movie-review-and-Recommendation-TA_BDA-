# Genuine n-gram language model
Run from this notebook folder after prepare_imdb.py. This experiment estimates token probabilities; it is not a sentiment classifier.

```python
from ngram_lab import NGram
toy = NGram(2).fit(["the movie is good", "the story is good", "the movie is bad"])
print(toy.next_words("the movie"))
print("Toy perplexity:", toy.perplexity(["the movie is good"]))
```
