## Optional export of measured comparison

Run only after selecting comparable actual reports. No missing value is replaced by zero. This plot does not select or retune a final-test model. Matplotlib must already be installed manually if used.


```python
if RUN_APPROVED_EXPERIMENTS and STUDY_IDS:
    values = [row for row in table if row['value'] is not None]
    if values:
        import matplotlib.pyplot as plt
        from validation.io import run_directory
        figure, axis = plt.subplots(figsize=(8, 4))
        axis.bar([row['study'] for row in values], [row['value'] for row in values])
        axis.set_ylabel(METRIC)
        axis.set_title('Comparable actual held-out reports')
        figure.tight_layout()
        folder = run_directory('phase13')
        figure.savefig(folder / 'comparison.svg')
        plt.show()
```
