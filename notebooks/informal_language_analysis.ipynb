```python
CONFIG_PATH = ROOT / "ml/multilingual_sentiment/config.yaml"
if RUN_APPROVED_EXPERIMENTS:
    from collections import Counter
    from ml.common.config import load_config
    from ml.common.data import load_experiment
    from ml.informal_language.preprocess import observations, feature_view
    bundle = load_experiment(load_config(CONFIG_PATH))
    counts = Counter()
    for record in bundle.records["train"][:128]:
        text = record.read_text()
        view = feature_view(text)
        details = observations(text)
        assert view.original == text
        counts["reviews"] += 1
        counts["symbol_runs"] += len(details["symbols"])
        counts["multiple_scripts"] += details["scripts"]["contains_multiple_scripts"]
    print(dict(counts))  # Counts only, not review text.
else:
    print("No text inspected; observations do not imply a trained model.")
```
